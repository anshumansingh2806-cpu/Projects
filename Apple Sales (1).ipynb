```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
df= pd.read_csv('apple_products.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

# Top 10 highest starrated iphone

```python
highest_rated= df.sort_values(by = ['Star Rating'],ascending = False)
highest_rated=highest_rated.head(10)
print(highest_rated['Product Name'])
```

# How many ratings does the highest rated iPhones have on flipkart?

```python
iphones = highest_rated['Product Name'].value_counts()
label=iphones.index
counts=highest_rated['Number Of Ratings']
fig=px.bar(highest_rated,x=label,y=counts,title = "Highest number of ratings on Flipkart")
fig.show()
```

# Which iPhone has the highest number of reviews on Flipkart?

```python
iphones = highest_rated['Product Name'].value_counts()
label=iphones.index
counts=highest_rated['Number Of Reviews']
fig=px.bar(highest_rated,x=label,y=counts,title = "Highest number of reviews on Flipkart")
fig.show()
```

# What is the relationship between sale price of iPhone and the number of ratings on Flipkart?

```python
figure= px.scatter(data_frame=df,x='Number Of Ratings',
                   y='Sale Price', size = 'Discount Percentage',trendline='ols',
                   title="Relationship Between Number of Ratings of iPhone and Sales Price")
figure.show()
```

# Relationship between Discount Percentage and Number of Ratings

```python
figure= px.scatter(data_frame=df, x='Number Of Ratings',
                   y='Discount Percentage',size='Sale Price',trendline='ols',
                   title="Relationship between NUmber of ratings in Flipkart and Discount Percentage")
figure.show()
```

# Most Expensive and Least Expensive

```python
most=df.loc[df['Sale Price'].idxmax()]
least=df.loc[df['Sale Price'].idxmin()]
print('Most Expensive Product',most)
print('Least Expensive Product',least)
```
