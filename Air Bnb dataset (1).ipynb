```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Air BNB.csv')
```

```python
df.head()
```

# Checking Missing Values

```python
print(df.isnull().sum())
```

```python
df.info()
```

```python
df['last review']=pd.to_datetime(df['last review'], errors ='coerce')
```

```python
df.fillna({'reviews per month' : 0, 'last review' : df['last review'].min()}, inplace = True)
```

```python
df.dropna(subset = ['NAME' , 'host name'],inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df=df.drop(columns=['license','house_rules'],errors='ignore')
```

```python
## Replacing dollar sign and converting the datatype as float

df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.describe()
```

# Visualisation

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='red')
plt.title('Distribution of listing price')
plt.xlabel("Price $")
plt.ylabel('Frequency')
plt.show()
```

# How diffrent room types are distributed?

```python
plt.figure(figsize=(8,5))
sns.countplot(x='room type',data=df,color='blue')
plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.countplot(y='neighbourhood group',data=df,color='green' , order=df['neighbourhood group'].value_counts().index)
plt.title('Number of listings by the neighbourhood')
plt.xlabel('Count')
plt.ylabel('Neighbourhood Group')
plt.show()
```

# Realtionship between price and room type?

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='room type', y='price', hue='room type', data=df, palette='Set1')
plt.title('Room Type vs Price')
plt.xlabel("Room Type")
plt.ylabel("Price in dollars")
plt.legend(title='Room Type')
plt.show()
           
```

# How has the number of reviews changed over time?

```python
df['last review']=pd.to_datetime(df['last review'])
reviews_change =df.groupby(df['last review'].dt.to_period('Y')).size()

plt.figure(figsize=(12,6))
reviews_change.plot(kind='bar',color='red')
plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.show()
```
